```python
data1 = []
print("請輸入A倉文具，輸入'/'結束：")
while True:
    item = input()
    if item == "/":
        break
    data1.append(item)

data2 = []
print("請輸入B倉文具，輸入'/'結束：")
while True:
    item = input()
    if item == "/":
        break
    data2.append(item)
result = data1 + data2 + ["膠水"]
print("二倉合併清單：")
print("/".join(result))
```
