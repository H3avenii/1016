```python
def calc(價格, 數量):
    小計 = 價格 * 數量
    稅金 = 小計 * 0.05
    return 小計, 稅金, 小計 + 稅金

p = int(input("請輸入單價："))
q = int(input("請輸入數量："))
小計, 稅金, 總金額 = calc(p, q)

print(f"小計：{小計}")
print(f"稅金：{稅金}")
print(f"總金額：{總金額}")
```
